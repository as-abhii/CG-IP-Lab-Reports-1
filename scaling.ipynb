```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('//content/apple.webp')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
h, w = image.shape[:2]
```

```python
# Scaling factors
sx = 3.5
sy = 5.5
```

```python
# Scaling matrix
M = np.float32([
  [sx, 0, 0],
  [0, sy, 0]
])
```

```python
scaled = cv2.warpAffine(
  image,
  M,
  (int(w * sx), int(h * sy))
)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(scaled)
plt.title("Scaled Image")
plt.axis("off")
plt.show()
```
