```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('//content/apple.webp')
image = cv2.cvtColor (image, cv2.COLOR_BGR2RGB)

#image dimension
h, w = image.shape[:2] #height, width, channel
#translation value
tx = 100
ty = 50

#translation
M=np.float32([
[1, 0, tx],
[0, 1, ty]
])
```

```python
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title('Original')
plt.axis('off')

# Define 'translated' image here by applying the affine transformation
translated = cv2.warpAffine(image, M, (w, h))

plt.subplot(1, 2, 2)
plt.imshow(translated)
plt.title('Translated')
plt.axis('off')
plt.show()
```
