```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('//content/apple.webp')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
h, w = image.shape[:2]
```

```python
# Center of image
center = (w // 2, h // 2)
```

```python
# Rotation angle
angle = 75
```

```python
# Rotation matrix
M = cv2.getRotationMatrix2D(center, angle, 1.0)
rotated = cv2.warpAffine(
    image,
    M,
    (w, h)
)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(rotated)
plt.title("Rotated Image - 75°")
plt.axis("off")
plt.show()
```
