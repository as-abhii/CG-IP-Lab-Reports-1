```python
import cv2
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('//content/apple.webp')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
# Reflection about X-axis
reflected_x = cv2.flip(image, 0)
reflected_y = cv2.flip(image, 1)
```

```python
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(reflected_x)
plt.title("Reflection about X-axis")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(reflected_y)
plt.title("Reflection about Y-axis")
plt.axis("off")

plt.show()
```
